```python

"""Question 2.Write a Python function that converts temperature from Fahrenheit to Celsius. Include a docstring, type hints, and input validation. Test it with three values including one invalid input.

    Answer;"""

def fahrenheit_to_celsius(fahrenheit: float) -> float:
    """
    Converts temperature from Fahrenheit to Celsius.
    """

    if isinstance(fahrenheit, bool) or not isinstance(fahrenheit, (int, float)):
        raise TypeError("Input must be a number (int or float).")

    celsius = (fahrenheit - 32) * 5 / 9
    return celsius


# Test cases
print(f"32 F is {fahrenheit_to_celsius(32):.2f} C")
print(f"212 F is {fahrenheit_to_celsius(212):.2f} C")

try:
    print(f"'abc' F is {fahrenheit_to_celsius('abc')} C")
except TypeError as e:
    print(f"Error with invalid input: {e}")
```

Question 5.   What is the output of barlow_wall_thickness(4467, 10.75, 65000)? Show the manual calculation step by step to verify your answer

Answer;
To determine the wall thickness using Barlow's Formula, we first state the standard relationship between pressure, wall thickness, outside diameter, and allowable stress:

$$P = \frac{2 \cdot S \cdot t}{D}$$

Rearranging the formula to solve for the minimum nominal wall thickness ($t$):

$$t = \frac{P \cdot D}{2 \cdot S}$$

---

### Given Parameters

* **Internal Design Pressure ($P$):** $4{,}467 \text{ psi}$
* **Outside Diameter ($D$):** $10.75 \text{ in}$
* **Allowable Stress ($S$):** $65{,}000 \text{ psi}$

---

### Step-by-Step Manual Calculation

1. **Calculate the numerator ($P \times D$):**

$$P \times D = 4{,}467 \times 10.75 = 48{,}020.25$$


2. **Calculate the denominator ($2 \times S$):**

$$2 \times S = 2 \times 65{,}000 = 130{,}000$$


3. **Divide the numerator by the denominator ($t = \frac{P \cdot D}{2 \cdot S}$):**

$$t = \frac{48{,}020.25}{130{,}000} = 0.36938653846... \text{ in}$$



---

The calculated wall thickness using Barlow's formula for the given values is approximately **0.3694 inches** (or $0.369 \text{ in}$).


```python
"""Question 6.Write a Python class called SubseaPipeline with properties for length_km, od_inches, wt_mm, material_grade, and a method that calculates the pipe volume in cubic metres.

Answer;"""
import math

class SubseaPipeline:
    def __init__(self, length_km, od_inches, wt_mm, material_grade):
        self.length_km = length_km
        self.od_inches = od_inches
        self.wt_mm = wt_mm
        self.material_grade = material_grade

    def pipe_volume_m3(self):
        # Convert units
        length_m = self.length_km * 1000
        od_m = self.od_inches * 0.0254
        wt_m = self.wt_mm / 1000

        # Calculate internal diameter
        id_m = od_m - (2 * wt_m)

        if id_m <= 0:
            raise ValueError("Wall thickness is too large for the pipe diameter.")

        # Calculate internal volume
        volume = math.pi * (id_m / 2) ** 2 * length_m

        return volume


# Create a pipeline object
pipeline1 = SubseaPipeline(
    length_km=10,
    od_inches=12,
    wt_mm=20,
    material_grade="X65"
)

# Display pipeline properties
print("Material grade:", pipeline1.material_grade)
print("Pipeline length:", pipeline1.length_km, "km")
print("Outside diameter:", pipeline1.od_inches, "inches")
print("Wall thickness:", pipeline1.wt_mm, "mm")

# Calculate and display volume
print("Internal pipe volume:", round(pipeline1.pipe_volume_m3(), 2), "m³")
```

```python
Question 9. Explain the try/except pattern. 9b Write a function that safely divides two numbers and returns ‘Error: Division by zero’ if the denominator is zero

Answer;
The try/except pattern in Python is used for exception handling. It allows a program to handle errors without stopping unexpectedly.

try: Contains code that might cause an error.

except: Runs when a specified error occurs.

return: Sends a result or error message back from a function.
```

```python
"""9b Write a function that safely divides two numbers and returns ‘Error: Division by zero’ if the denominator is zero
Answer;"""

def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "Error: Division by zero"


# Test cases
print(safe_divide(10, 2))
print(safe_divide(10, 0))
print(safe_divide(20, 4))
```
