```python
"""QUESTION ONE (1) Write a function reservoir_temperature(depth_m, surface_temp=4.0, gradient=30.0) that returns temperature in °C. Include docstring, type hints, and input validation."""


def reservoir_temperature(depth_m: float, surface_temp: float = 4.0, gradient: float = 30.0) -> float:
    """
    Calculates the reservoir temperature at a given depth.

    Args:
        depth_m (float): The depth in meters.
        surface_temp (float, optional): The surface temperature in degrees Celsius.
                                        Defaults to 4.0.
        gradient (float, optional): The geothermal gradient in degrees Celsius per kilometer.
                                    Defaults to 30.0.

    Returns:
        float: The temperature in degrees Celsius at the specified depth.

    Raises:
        TypeError: If any input is not numeric (float or int).
        ValueError: If depth_m is negative or gradient is not positive.
    """
    # Input validation for numeric types
    if not isinstance(depth_m, (int, float)):
        raise TypeError("Depth (depth_m) must be a numeric value.")
    if not isinstance(surface_temp, (int, float)):
        raise TypeError("Surface temperature (surface_temp) must be a numeric value.")
    if not isinstance(gradient, (int, float)):
        raise TypeError("Gradient (gradient) must be a numeric value.")

    # Input validation for value ranges
    if depth_m < 0:
        raise ValueError("Depth (depth_m) cannot be negative.")
    if gradient <= 0:
        raise ValueError("Geothermal gradient (gradient) must be positive.")

    # Convert depth from meters to kilometers and gradient from °C/km
    depth_km = depth_m / 1000.0
    temperature = surface_temp + (gradient * depth_km)
    return temperature
```

```python
"""QUESTION THREE (3)  Create a scatter plot showing temperature vs. depth with a linear trendline. Label axes with units, add a title, and include the trendline equation in the legend."""


import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import linregress

def reservoir_temperature(depth_m: float, surface_temp: float = 4.0, gradient: float = 30.0) -> float:
    """
    Calculates the reservoir temperature at a given depth.

    Args:
        depth_m (float): The depth in meters.
        surface_temp (float, optional): The surface temperature in degrees Celsius.
                                        Defaults to 4.0.
        gradient (float, optional): The geothermal gradient in degrees Celsius per kilometer.
                                    Defaults to 30.0.

    Returns:
        float: The temperature in degrees Celsius at the specified depth.

    Raises:
        TypeError: If any input is not numeric (float or int).
        ValueError: If depth_m is negative or gradient is not positive.
    """
    # Input validation for numeric types
    if not isinstance(depth_m, (int, float)):
        raise TypeError("Depth (depth_m) must be a numeric value.")
    if not isinstance(surface_temp, (int, float)):
        raise TypeError("Surface temperature (surface_temp) must be a numeric value.")
    if not isinstance(gradient, (int, float)):
        raise TypeError("Gradient (gradient) must be a numeric value.")

    # Input validation for value ranges
    if depth_m < 0:
        raise ValueError("Depth (depth_m) cannot be negative.")
    if gradient <= 0:
        raise ValueError("Geothermal gradient (gradient) must be positive.")

    # Convert depth from meters to kilometers and gradient from °C/km
    depth_km = depth_m / 1000.0
    temperature = surface_temp + (gradient * depth_km)
    return temperature


# Create a dummy CSV file with 50 random depth measurements if it doesn't exist
# This ensures the example is runnable independently.
try:
    df_depths = pd.read_csv('well_depths.csv')
    print("Using existing 'well_depths.csv' file.")
except FileNotFoundError:
    np.random.seed(42) # for reproducibility
    well_depths = np.random.uniform(500, 5000, 50)
    df_depths = pd.DataFrame({'depth_m': well_depths})
    df_depths.to_csv('well_depths.csv', index=False)
    print("Generated 'well_depths.csv' with 50 random depth measurements.")

# Apply the reservoir_temperature function to the 'depth_m' column to create wells_df
wells_df = df_depths.copy() # Create a copy to store results
wells_df['temperature_celsius'] = wells_df['depth_m'].apply(reservoir_temperature)

# Perform linear regression to get the trendline equation
slope, intercept, r_value, p_value, std_err = linregress(wells_df['depth_m'], wells_df['temperature_celsius'])
trendline_eq = f'y = {slope:.4f}x + {intercept:.2f} (R^2 = {r_value**2:.2f})'

# Create the scatter plot with a linear regression line
plt.figure(figsize=(10, 6))
sns.regplot(x='depth_m', y='temperature_celsius', data=wells_df, scatter_kws={'alpha':0.6}, line_kws={'color':'red'})

# Add title and labels with units
plt.title('Reservoir Temperature vs. Well Depth')
plt.xlabel('Well Depth (meters)')
plt.ylabel('Temperature (°C)')

# Display the trendline equation in the legend
plt.legend(title=trendline_eq, loc='upper left', frameon=True)

plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
